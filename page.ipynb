```python
import cv2
import os
import pytesseract
import face_recognition as fr
import numpy as np
from datetime import datetime
import csv
```

```python
path=r'C:\Users\91859\Deep_learning\attendence_system\students'
```

```python
mylist=os.listdir(path)
mylist
```

```python
imgs=[]
classnames=[]

for i in mylist:
    imagepath=os.path.join(path,i)
    current_image=cv2.imread(imagepath)
    imgs.append(current_image)
    classnames.append(i.split('.')[0])
```

```python
encodelist=[]

def encoding_images(images):
    for img in images:
        img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
        face_in_frame=fr.face_locations(img)
        encode_face=fr.face_encodings(img,face_in_frame)[0]
        encodelist.append(encode_face)
    return encodelist
    

known_faces=encoding_images(imgs)
```

```python
def mark_attendance(name):
    file_name='attendence.csv'
    
    now=datetime.now()
    date=now.strftime('%d-%m-%Y')
    time=now.strftime('%H-%M-%S')
    
    
    if not os.path.exists(file_name):
        with open(file_name,'w',newline='') as f:
            writer=csv.writer(f)
            writer.writerow(['Name','Date','Time'])
            
    with open(file_name,'r',newline='')as f:
        reader=csv.reader(f)
        next(reader,None)
        
        for row in reader:
            if row[0]==name and row[1]== date:
                return
            
            
    with open(file_name,'a',newline='') as f:
        writer=csv.writer(f)
        writer.writerow([name,date,time])
        
    print(f"{name} marked at {time}")
```

```python
video=cv2.VideoCapture(0)
while True:
    succ,img=video.read()
    img1=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    face_loc=fr.face_locations(img1)
    face_encode=fr.face_encodings(img1,face_loc)
    
    
    for enc_face,loc_face in zip(face_encode,face_loc):
        math=fr.compare_faces(known_faces,enc_face)
        
        face_dist=fr.face_distance(known_faces,enc_face)
        
        
        match_index=np.argmin(face_dist)
        
        if math[match_index]:
            name=classnames[match_index]
            
            mark_attendance(name)
            
            # print(name)
            y1,x2,y2,x1=loc_face # x1,y1 -> left top  -> this be like displayed
            cv2.rectangle(img,(x1,y1),(x2,y2),(0,0,0),3)
            cv2.putText(img,name,(x1,y1),cv2.FONT_HERSHEY_COMPLEX,2,(0,0,200))
            
        else:
            y1,x2,y2,x1=loc_face # x1,y1 -> left top  -> this be like displayed
            cv2.rectangle(img,(x1,y1),(x2,y2),(0,0,0),3)
            cv2.putText(img,'Unknown_Person',(x1,y1),cv2.FONT_HERSHEY_COMPLEX,2,(0,0,200))
            
            
    cv2.imshow('People',img)
    if cv2.waitKey(1) & 0XFF==ord('q'):
        break
video.release()
cv2.destroyAllWindows()
            
```
